## The question

> mirror aka does e_0-e_7 mirrored (non chirally reversed) across the boundary?"*

Cayley-Dickson doubling builds each algebra from two copies of the last:

```
(a, b) (c, d)  =  (ac - d* b,   d a + b c*)
```

R -> C -> H -> O -> S(16) -> T32(32) -> 64 -> 128 ...

Each step draws a **boundary** halfway up the index range. The question is what
kind of object that boundary is.


## Why it is not a decorative question

Three separate threads land on the same boundary.

**1. Provenance.** `Operating-L-IO.md` needs to distinguish an internal signal from
an external one. From inside the machine that is undecidable — function scope and
the equation being evaluated are the same object, so a decider can be diagonalised.
The only remaining move is to read *which coordinates lit up*, which makes the
structure of the index space the whole apparatus.

**2. The orphans.** Phase 25 found that `e_0` and `e_8` sit in **no Assessor** — the
two directionless channels, the part of the structure whose content is its own
absence. Nobody had asked why those two, or whether the pattern continues.

**3. The reflection.** Phase 27.7 derived `sigma -> 1 - sigma` as a reflection across
a diameter with **exactly two fixed points, sigma = 0 and sigma = 1/2.** Two fixed
points, two orphans. That coincidence is either meaningless or it is the claim.


## The lever

A lever needs a fulcrum, and a fulcrum does no work. It is the one point that does
not move while everything else balances across it.

In 0_RB the null subspace is described exactly that way — **gravity, present as
absence**: dimensions that exist in the operator, are indexed, participate in the
algebra, and return zero. Not unused capacity.

If the boundary is a mirror, its fixed points are the fulcrum. If the counts on
either side match, it balances. Both halves of that are computable, so neither
needs to be argued.

## What this paper does NOT claim

⚠ Stated up front so it cannot be overclaimed later.

- **Nothing about observers.** "One sedenion per observer" is an interpretive
  assumption. The algebra can rule a symmetry in or out; it cannot tell you what
  the halves *are*.
- **Nothing about meaning.** That needs a source position `beta` and no computation
  here supplies one.
- **Nothing about physics.** The {4,8,4} force reading is quoted from 0_RB as
  context, not re-derived or defended.
