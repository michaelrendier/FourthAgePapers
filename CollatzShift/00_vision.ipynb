## The three things worth knowing before the measurements

**1. The recursion is real and it is on the modulus.** `T^k` restricted to a
residue class mod `2^k` is an *affine map* — one multiplication and one addition.
There are `2^k` of them and refining `k` to `k+1` splits every branch in two.
The "modular function" is the indexing of the pieces by `Z/2^k`, and the
"recursive decomposition" is that refinement.

**2. The dynamics are completely solved.** The parity word — the sequence of
even/odd decisions along an orbit — is a *2-adic address*, and `T` acts on it by
**chopping off one digit**. Collatz is the binary shift map in disguise.

**3. So the difficulty is not the dynamics.** The shift has entropy `log 2`,
dense periodic points, uncountably many orbits — total chaos, fully understood.
The Collatz conjecture asserts that `N`, a dense measure-zero subset of `Z_2`,
**misses all of it**. The problem is not about the map. It is about how the
integers sit inside the 2-adics — a statement joining two incompatible metrics.
