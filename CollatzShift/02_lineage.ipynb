## The four-part test, applied

1. **Is it a count or a ratio of something else?** The map itself is — it is a
   composition — so `T` lands at tier 3, DERIVED. Its *parts* are not.
2. **Is it a fixed set?** No. The fixed set of `T` (`{0, −1}` in `Z_2`) is an
   object the map *has*, not the map.
3. **Does it change length?** Both branches do — gains `1/2` and `3`. So both
   need **DILATE** (tier 1), not merely REFLECT.
4. **Does it need an added constraint to exist?** No. `T` is total on `Z_2`.

## Verdict: no new generator

`T = ADD ∘ SCALE ∘ SIGN`. All three tier-0 irreducibles, and nothing else.

**And that is exactly why it is hard.** The Collatz map is the *smallest
non-trivial composition that uses all three tier-0 primitives at once*. ADD and
SCALE have **different identities** — `0` and `1`. The `+1` is the only thing
coupling the two axes, and every difficulty in the problem lives in that
coupling.

Remove it and the difficulty vanishes with it:

```python
# Drop the +1. What is left is a pure SCALE tower, and it is trivial.
def T_no_add(n):
    while n % 2 == 0:
        n //= 2
    return 3 * n

n = 7
print('n -> 3n/2^v, the +1 removed (pure SCALE, no ADD coupling):')
for _ in range(8):
    n = T_no_add(n)
    print('   ', n)
print('\n3n with n odd is odd, so v = 0 forever: strictly ×3 each step.')
print('No conjecture in it. The +1 is the whole problem.')
```

**No new generator required.**

## Where the framework *does* speak — labelled as structure, not identity

Four correspondences to existing machinery in this project. The first four are
structural identities; the fifth is a correction to an analogy that does **not**
hold, recorded so it does not get made later.

**`VAPMIP/intention_monad.py` — the conjecture is the GC completeness claim.**
Reachability is the image, collection is the kernel, the root set is the
intention, and `ker(AB) ⊇ ker(B)` makes a run a monotone descent. Set the root
to `{1}`: the Collatz conjecture says **run the collector on `N` and it collects
nothing** — the kernel is empty. The descent is monotone in exactly that sense:
one parity bit is consumed per step and never restored (R8).

**`VAPMIP/sieve_clock.py` — the orphans are already there, with one difference.**
`n` has an odd predecessor iff `n ≡ 2 (mod 3)` (R7), so two thirds of `N` are
leaves of the backward tree. That is the ORPHAN regime. But a sieve orphan at `N`
is *adopted at `2N`* — its bound is the universe size. A Collatz orphan is
**never** adopted; its bound is algebraic. Two kinds of orphan, and the
instrument currently only models one.

**`L_(I|O)` — the same inside-out asymmetry.** Forward is deterministic and free.
Backward is a branching tree gated by a `mod 3` test. Forward throws away one bit
per step; backward has to guess it. **The parity word is the private key** — the
same shape as *speaking English is easy, the inverse is factoring the modulus*.

**`0_RB = e_0` — the attractor is the identity.** Collatz's fixed points in `Z_2`
are the two constant words; on `N` the multiplicative identity is reachable only
as an ordered 2-cycle. "Every trajectory falls into the identity" is the
conjecture in this project's vocabulary.

**⚠ The Two Trees do NOT split even/odd.** Telperion = prime, Laurelin =
composite. Even/odd is divisibility by 2, not primality — `2` is prime and even,
`9` is composite and odd. The tempting mapping is **wrong** and is recorded here
as wrong. The defensible Two Trees reading of Collatz is at the *tower* level:
the 2-tower (contracting, `×1/2`) against the 3-tower (expanding, `×3/2`),
mingling at `log₂3 = 1.58496`, which is where every candidate cycle must live
(R6).
