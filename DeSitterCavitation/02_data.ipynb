## The matching condition (input, not a result)

Gravastar match — interior de Sitter `1 − (r/L)²` joins exterior Schwarzschild `1 − r_s/r` at the shell `r ≈ r_s` ⇒ `L_dS = r_s`. This is the assumption; the paper's claim is what follows from it.

```python
Msun = M.M_SUN
print('CALIBRATION: L_dS == r_s by construction, across the mass range')
for e in range(-6, 10, 2):
    Mk = (10.0**e) * Msun
    print(f'  M = 1e{e:+03d} Msun   r_s = {M.r_schwarzschild(Mk):.4e} m   '
          f'L_dS = {M.L_desitter(Mk):.4e} m')
```
