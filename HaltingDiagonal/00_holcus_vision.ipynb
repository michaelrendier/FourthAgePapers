## The computer science

The undecidability of the halting problem is usually taught as a proof by
contradiction, and the contradiction is where attention goes. This paper puts
attention on the *construction* instead.

Given a table `T[i][j]` recording whether program `Pᵢ` halts on input `Pⱼ`, the
diagonal program is

```
D[i] = 1 - T[i][i]
```

`D` differs from row `i` at column `i`, for every `i`. Therefore `D` is not any
row of `T`. That is the whole engine of the proof, and it has three properties
worth stating in CS terms:

1. **It is constructive.** `D` is not shown to exist; it is built, in one pass.
2. **It is decidable, and cheaply.** Checking that `D` escapes costs `O(n)` —
   and it depends *only on the diagonal* of `T`, never on the other `n² − n`
   entries. So exhaustive verification over all tables of order `n` costs
   `2ⁿ`, not `2^(n²)`.
3. **It is an involution.** `x ↦ 1 − x` on bits, `x ↦ −x` on signs. Applying it
   twice is the identity. Applying it once has no fixed point.

The last property is the one that unifies. A permutation with no fixed point is
a **derangement**. The Enigma reflector was wired to be exactly that — no letter
could ever encrypt to itself — and that constraint is what made the Bombe's crib
attack tractable. The same no-fixed-point condition is what makes Cantor's
diagonal escape its list, and what makes `D(D)` contradict any halting oracle.

## What is being claimed, and what is not

**Claimed:** the four constructions are the same operation; the escape property
is machine-checkable; the derangement count is exact; the no-fixed-point
constraint yields a measurable search-space reduction.

**Not claimed:** that this decides the halting problem. It does not, and nothing
here should be read as suggesting otherwise. HALT remains undecidable. What is
shown is that the *step which proves it undecidable* is finite, constructive and
verifiable — an observation about proof structure, not about computability.
