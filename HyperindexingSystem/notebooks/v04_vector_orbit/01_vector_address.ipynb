# v04 · VectorAddress (uint32 limbs)
`infinite-data-vector-shaders.py` — the integer address as a little-endian array of 32-bit words, sized for a GPU buffer.

```python
import importlib.util, os, sys, time, math, hashlib, json
from collections import Counter
PROJ = os.environ.get("PROJ", "/home/rendier/Projects/ThePlace")
NB = os.path.join(PROJ, "FourthAgePapers/HyperindexingSystem/notebooks")
CORPUS = open(os.path.join(NB, "corpus_sample.txt"), encoding="utf-8").read()   # fixed 4000-char sample, 97-symbol keyboard set
def load(rel, name="m"):
    """Load a source file from the repo by path, unmodified."""
    spec = importlib.util.spec_from_file_location(name, os.path.join(PROJ, rel))
    m = importlib.util.module_from_spec(spec); spec.loader.exec_module(m); return m
print("corpus:", len(CORPUS), "chars, sha256", hashlib.sha256(CORPUS.encode()).hexdigest()[:16])
```

```python
vs = load("PtolemyDesktop/Callimachus/HyperWebster-Data-Storage/infinite-data-vector-shaders.py", "vs")
VA = vs.VectorAddress
B = vs._Bijection()
for size in (13, 140, 1000, 4000):
    a = B.str_to_int(CORPUS[:size]); va = VA.from_integer(a, size)
    assert va.to_integer() == a and VA.from_label(va.to_label(), size).to_integer() == a
    print("%5d chars  %6d bits  %4d words  %3d uvec4  label %d chars" % (size, a.bit_length(), va.num_words, va.num_uvec4s, len(va.to_label())))
print(VA.from_integer(0xdeadbeefcafe, 1).to_label())
```

Every 32-bit word is written as 8 hex digits and joined by `_`; the label is 9 characters per word, so it is **larger** than the plain hex address it encodes (8 digits per word plus a separator).

```python
va = VA.from_integer(B.str_to_int(CORPUS[:1000]), 1000)
plain_hex = format(va.to_integer(), "x")
print("plain hex digits:", len(plain_hex), "| label chars:", len(va.to_label()))
```
