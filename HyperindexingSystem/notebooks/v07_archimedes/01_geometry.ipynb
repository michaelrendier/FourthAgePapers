```python
import importlib.util, os, sys, time, math, hashlib, json
from collections import Counter
PROJ = os.environ.get("PROJ", "/home/rendier/Projects/ThePlace")
NB = os.path.join(PROJ, "FourthAgePapers/HyperindexingSystem/notebooks")
CORPUS = open(os.path.join(NB, "corpus_sample.txt"), encoding="utf-8").read()   # fixed 4000-char sample, 97-symbol keyboard set
def load(rel, name="m"):
    """Load a source file from the repo by path, unmodified."""
    spec = importlib.util.spec_from_file_location(name, os.path.join(PROJ, rel))
    m = importlib.util.module_from_spec(spec); spec.loader.exec_module(m); return m
print("corpus:", len(CORPUS), "chars, sha256", hashlib.sha256(CORPUS.encode()).hexdigest()[:16])
```

```python
ar = load("PtolemyDesktop/Callimachus/HyperWebster-Data-Storage/hyperwebster_archimedes.py", "ar")
AG = ar.ArchimedesGeometry
# locate the maximum of V(n) = pi^(n/2) / Gamma(n/2 + 1) by golden-section search
f = lambda n: AG.log_vol(n)
a, b = 3.0, 8.0; g = (math.sqrt(5) - 1) / 2
c, d = b - g*(b-a), a + g*(b-a)
for _ in range(80):
    if f(c) > f(d): b = d
    else: a = c
    c, d = b - g*(b-a), a + g*(b-a)
peak = (a + b) / 2
print("numerical peak of the unit n-ball volume: n =", round(peak, 4))
print("code's peak_dimension() returns        :", round(AG.peak_dimension(), 4), "(= 2π − 1)")
```

The module carries three different statements of the peak: the docstring's 'n~5', the method docstring's '~4.93', and `2π − 1 = 5.283` in the return value. The measured maximum is 5.2569. The return value is off by 0.026, the method docstring by 0.32.

```python
for n in (5, 5.2569, 97):
    print(n, "log10(ball/cube) =", round(AG.log_fraction_of_cube(n), 3))
print("readable_fraction_estimate(1000):", round(AG.readable_fraction_estimate(1000), 1), "(log10, assumes 4.5 bits/char English entropy)")
```
