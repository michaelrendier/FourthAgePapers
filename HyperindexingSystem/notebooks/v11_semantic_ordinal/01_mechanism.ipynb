# v11 · Semantic address: surface form → Riemann-zero ordinal
`SemanticWordEngine/hyperwebster.py` (git 2026-05-12). Not the exact bijection: a many-to-one map from a word to one of N zeros.

```python
import importlib.util, os, sys, time, math, hashlib, json
from collections import Counter
PROJ = os.environ.get("PROJ", "/home/rendier/Projects/ThePlace")
NB = os.path.join(PROJ, "FourthAgePapers/HyperindexingSystem/notebooks")
CORPUS = open(os.path.join(NB, "corpus_sample.txt"), encoding="utf-8").read()   # fixed 4000-char sample, 97-symbol keyboard set
def load(rel, name="m"):
    """Load a source file from the repo by path, unmodified."""
    spec = importlib.util.spec_from_file_location(name, os.path.join(PROJ, rel))
    m = importlib.util.module_from_spec(spec); spec.loader.exec_module(m); return m
print("corpus:", len(CORPUS), "chars, sha256", hashlib.sha256(CORPUS.encode()).hexdigest()[:16])
```

```python
sw = load("SemanticWordEngine/hyperwebster.py", "sw")
monad = load("VAPMIP/monad.py", "monad")
zeros = {k: v for k, v in monad._ZERO_CACHE.items() if k <= 50}          # first 50 exact zeros
lex = sw.Lexicon(zeros, path="/dev/null")
for w in ["listen", "silent", "enlist", "tinsel", "Listen", "hello"]:
    print("%-8s _hash = %-28s address = %s" % (w, sw._hash(w), lex.address_of(w)))
```

`_hash` is `(mean token length, (Σ ord) mod 360 / 360)`. A sum is unchanged by any permutation of the characters, so **every anagram of a word has the same address**: the map is constant on each orbit of the position-permutation group. That is the opposite of the exact address (v02), which separates every string.

```python
assert lex.address_of("listen") == lex.address_of("silent") == lex.address_of("enlist") == lex.address_of("tinsel")
import re
words = sorted(set(re.findall(r"[a-z]+", CORPUS.lower())))
addrs = Counter(lex.address_of(w)[0] for w in words)
ts = {round(sw._hash(w)[1], 6) for w in words}
print("distinct words:", len(words), "| distinct t values:", len(ts), "(at most 360) | distinct zero ordinals:", len(addrs), "of 50 | largest bucket:", max(addrs.values()))
assert len(ts) <= 360
```

With 50 zeros the address space has 50 cells; the corpus words are spread over them and many share a cell by construction. The file's own `N_ZEROS` is 200. A many-to-one map is the intended behaviour of a *semantic* address (synonyms share a zero); the anagram invariance is a side effect of the choice of `Σ ord`.
