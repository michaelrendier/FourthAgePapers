# v15 · Sorted rotations of a string: the Burrows–Wheeler transform
Burrows & Wheeler (SRC-RR-124, 1994; read in full): 'forms the N rotations of S, sorts them lexicographically, extracts the last character'. The rotations of a string are the rows of a tabula recta laid over the string. Reproduced against the paper's own worked example.

```python
def bwt(s):
    rows = sorted(s[i:] + s[:i] for i in range(len(s)))
    return "".join(r[-1] for r in rows), rows.index(s)
def ibwt(L, I):
    n = len(L); table = [""] * n
    for _ in range(n):
        table = sorted(L[i] + table[i] for i in range(n))
    return table[I]
L, I = bwt("abraca")
print("L =", L, " I =", I)
assert (L, I) == ("caraab", 1)               # the paper's example: L = 'caraab', I = 1
assert ibwt(L, I) == "abraca"
rows = sorted("abraca"[i:] + "abraca"[:i] for i in range(6)); print(rows)
```

The transform is a permutation of the input (`L` is a rearrangement of `abraca`) plus one index, and it is reversible. It is the same object as v15/01's tabula recta rows, sorted.

```python
import random
rnd = random.Random(1)
for _ in range(200):
    s = "".join(rnd.choice("ab c") for _ in range(rnd.randint(1, 12)))
    Lx, Ix = bwt(s); assert ibwt(Lx, Ix) == s
print("200 random strings: bwt/ibwt round trip exact")
```
