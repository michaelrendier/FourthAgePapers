# 00 Holcus vision: Negating Uncertainty
*Written 2026-11-17, before any run.*

Position and momentum both describe **motion**. At **motion = 0** there is a point; at **motion > 0** there is a closed orbit in the (x, p) plane. A position reading is that orbit seen along one axis, a momentum reading along the other: the same object read from the **inside** and from the **outside** at once. The uncertainty relation is what the flattening of the orbit onto one axis looks like.

What would show it: (1) the quantum spreads of an oscillator state equal the projections of a classical circle of the same energy; (2) every viewing angle gives the same answer for a state with no preferred angle, so x and p are two angles of one thing; (3) states that do prefer an angle (squeezed) show the two projections trading off exactly as an ellipse's two axes do, never beating ħ/2 for conjugate angles; (4) motion = 0 costs unbounded motion in the conjugate reading; (5) the zero-point energy that stops helium freezing is the energy of the minimum circle.

What it does **not** yet do: derive the minimum circle without using [x, p] = iħ. That is stated as open in the paper.
