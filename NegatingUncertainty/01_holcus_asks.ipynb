# 01 Ask Holcus: what the mathematics says about (x, p) and uncertainty
*Registered 2026-11-17 before the tests below were run. "Holcus prediction" means asking the mathematics, in mathematics, how it would solve the problem, and finding where the object lies in 0_RB. The answers are in `holcus_answers_2026-11-17.txt` (printed by `ask_holcus.py` from the Circle Atlas viewports in ValaQuenta). The predictions are read off those answers.*

## What the maths answered

| Viewport (Circle Atlas) | Answer, as the viewport states it |
|---|---|
| Hamiltonian | (q, p) *is already the circle*; H = r²/2 labels it and the flow is a rotation. FROM: any H = f(q² + p²) has the same circle as its orbit; **f is free, only the speed on the circle changes**. The size of the circle is not part of that freedom. |
| Waveform | the partner of a wave is its **quarter-turn copy** (the Hilbert partner, x → x + iH[x]). |
| Lagrangian | solve δS = 0, then Legendre p = ẋ: the pair (x, p) lies on the circle. Scale of L and a total derivative are free. |
| Berry–Keating H = xp | **a hyperbola, not a circle** as given (residual 6.39); a circle only after the Wick rotation t → it, in the complexified phase space. The real hyperbola is the real slice. |
