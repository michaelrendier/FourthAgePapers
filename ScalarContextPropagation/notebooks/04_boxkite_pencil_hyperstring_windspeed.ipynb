## 1. The box kite, established

de Marrais (2000): 42 Assessors (planes `span(e_a, e_{b+8})`, `a,b in
1..7`, `a != b`), partitioned into 7 **box kites** by strut `s = a XOR b`,
6 Assessors each. Each box kite is the octahedron `K_{2,2,2}`: 4-regular,
12 edges, 3 non-edges (the **reversal pairs**, `(a,b) <-> (b,a)` — these
are the box kite's physical struts, the antipodal bars).

```python
s = 1
g = bk.box_kite_graph(s)
print(f"strut {s}: vertices={g['vertices']}")
print(f"  is_octahedron: {g['is_octahedron']}   4-regular: {g['is_4_regular']}")
print(f"  the 3 struts (antipodal/non-edge pairs): "
      f"{[(g['vertices'][i], g['vertices'][j]) for i,j in g['non_edges']]}")
print(f"  chart_spectrum: {bk.chart_spectrum(s)}   (expect {{0,4,4,4,6,6}})")
```

## 3. The Pencil HyperString — the construction, as conjectured

From `ValaQuenta/wiki/pencil_hyperstring.md` (status: **THEORETICAL** —
CONJECTURE, not yet settled):

> Fix strut `s`. Lay the seven factor-pairs of the pencil along `s in
> [0,1]` at seven equal stations; each station is a pure-imaginary
> generator `g_k` — the composition `Π = g_1 · g_2 · … · g_7` (sedenion
> product, order kept) has real part `H`, the string's conserved
> horizontal tension. Wind speed `w` sets a transverse pressure `q(w) ∝
> ½ρw²`; the loaded string's shape is a Joukowsky map at each strut
> crossing: `ζ ↦ ζ + (H/q(w))·ζ⁻¹`.

Inflation sequence (from the same source): `w=0` — string slack, all six
vertices collapse onto the e₀-strut axis, the box kite is a **point** (the
shadow); `0<w<w*` — sails catch, struts open, Assessors separate;
`w=w*` — regular octahedron, "the kite flies"; `w>w*` — over-pressure,
twisted box kite; `w>>w*` — the kite tears off the ZD surface. Some
charts have a range of `w` and deform through it (**deformable**); others
have one fixed form regardless of `w` (**non-deformable / rigid**).

## 4. Attempting `H = Re(Π)` literally — what actually happens

The construction names "a pure-imaginary generator per pencil station"
but does not say **which** unit direction represents a station `{a,b}`,
nor **what order** the seven stations compose in. Both choices are tried
below, honestly, against the sedenion product already verified in
`maths.py`.

```python
import functools

def re_of_product(reps):
    v = bk.basis_vector((1.0, reps[0]))
    for r in reps[1:]:
        v = bk.multiply(v, bk.basis_vector((1.0, r)))
    return v[0]

print("H = Re(Pi) for every strut, generator = min(a,b) per station, ascending order:")
for s in range(1, 8):
    stations = bk.pencil(s)
    reps = [min(a, b) for a, b in stations]
    print(f"  strut {s}: reps={reps}  xor={functools.reduce(lambda x,y: x^y, reps)}  H={re_of_product(reps)}")
```

```python
import random

print("Same strut (1), different (undocumented) modeling choices for 'which generator / which order':")
stations = bk.pencil(1)
reps_min = [min(a, b) for a, b in stations]
reps_max = [max(a, b) for a, b in stations]
reps_a   = [a for a, b in stations]
reps_b   = [b for a, b in stations]

print(f"  representative = min(pair), ascending order : H = {re_of_product(reps_min)}")
print(f"  representative = max(pair), ascending order : H = {re_of_product(reps_max)}")
print(f"  representative = first-listed element       : H = {re_of_product(reps_a)}")
print(f"  representative = second-listed element       : H = {re_of_product(reps_b)}")

random.seed(1)
shuffled_Hs = set()
for _ in range(20):
    r2 = reps_min[:]
    random.shuffle(r2)
    shuffled_Hs.add(re_of_product(r2))
print(f"  representative = min(pair), 20 random orders : H takes values {sorted(shuffled_Hs)}")
```

## `H`, measured — `THEORETICAL:CALCULATED`

Zero divisors are, in the established literature, a **pitfall** of
sedenion algebra — the thing that breaks the division-algebra property
octonions still have (Moreno 1997/98). Nothing here uses that algebra in
general — the 16 basis labels are used as a fixed combinatorial index
(which pairs vanish, which don't), not as operands closed under
open-ended sedenion multiplication. `H = Re(Π)`, composing seven such
placeholder generators, is measured exactly as `pencil_hyperstring.md`
specifies, under several explicit conventions (which member of each
factor pair, what order):

Every convention tried gives a **discrete** result — `Re(Π) ∈ {-1, 0,
1}` — never the continuously varying tension `Φ(w)` would need. That is
reported here as a measurement, not a defect: whether a given convention
lands the composition *on* the real axis or *off* it may itself be the
meaningful signal (an anchoring condition, in a reading this paper does
not develop), rather than noise to be engineered away. Resolving what the
discreteness means is a separate, deeper question, deliberately out of
scope here — this paper only needs to report, honestly, that `Φ(w)` is
not yet reducible to running code from this construction alone.

## 5. A toy scalar-to-structure interpolation (illustrative, NOT `Φ(w)`)

What *can* be shown honestly, with primitives that already ship, is the
**qualitative shape** of the inflation sequence — that a single scalar
sweeping from 0 to 1 can carry a state from "pure fixed point" to "fully
on one chart." This is a **linear** interpolation between `e₀` and the
(normalised) sum of the kite's six `d+` diagonal directions — not the
conjectured Joukowsky law, and not claimed to reconstruct individual
Assessor coordinates the way `Φ(w)` would. It demonstrates the shape of
the claim, not the claim.

```python
s = 1
kite = bk.box_kites()[s]
diag_sum = [0.0] * 16
for a, b in kite:
    d_plus = bk.diagonals(a, b)[0]
    for i in range(16):
        diag_sum[i] += d_plus[i]
n = bk.norm(diag_sum)
diag_unit = [c / n for c in diag_sum]

print(f"strut {s} kite: {kite}")
print("sweeping a toy scalar w in [0,1] between e0 and the kite's summed diagonal direction:")
for w in [0.0, 0.25, 0.5, 0.75, 1.0]:
    v = [(1 - w) * (1.0 if i == 0 else 0.0) + w * diag_unit[i] for i in range(16)]
    r = bk.chart_of(v)
    print(f"  w={w:.2f}  outside_share={r['outside_share']:.3f}  "
          f"dominant_chart={r['dominant_chart']}  chart_share={r['chart_share']:.3f}  "
          f"is_zero_divisor={r['is_zero_divisor']}")
```

`outside_share` (the fraction of the address's energy sitting on `e₀`,
i.e. "pure fixed point") runs 1.0 → 0.0 exactly as `w` runs 0 → 1, and the
kite's own strut (1) is correctly identified as the dominant chart the
moment `w > 0` — qualitatively the same shape the paper describes (`w=0`:
point/shadow; `w>0`: separates onto the chart). The interpolated state is
not itself a zero divisor at any `w` here (summing six different
diagonals is not one of the 84 ZD directions) — expected, and a genuine
difference from what a real `Φ(w)` would need to produce.
