# The Sonification Encoder — Holcus Vision

**One Claim:** The sonification map is **exact rational arithmetic** — all 30
named tones are exact `Fraction` just-intonation ratios of A440, so frequency
assignment introduces no rounding at all — but the encoder is **lossy in two
independent ways**: the map is not injective (30 names collapse onto 23 distinct
frequencies), and 4 of the 6 rest durations silently truncate under floor
division despite being labelled *"exact integer arithmetic"*.

**Data:** None external. Every value is a rational computed from a definition.

## The computer science

This module is an **encoder**. It takes symbols — particle names, algebra strata
— and assigns each one a frequency and a duration, producing an audio stream.
The only questions that matter about an encoder are:

1. **Is the encoding exact?** Does the arithmetic that produces a code introduce
   error?
2. **Is it injective?** Can two distinct symbols receive the same code?
3. **Is it decodable?** Given the stream, can the symbol sequence be recovered?

These are not aesthetic questions and they are not physics questions. They are
questions about a codec, and each has a definite answer that code can produce.

## What is found

**On exactness of frequency: the module is right, and better than it needs to
be.** `_jf(num, den, octave)` builds every frequency as
`Fraction(num, den) × 440`, in exact rational arithmetic. Nothing is computed as
`440 * 2**(k/12)`, which is the usual equal-tempered approach and is irrational
by construction. Every tone is therefore an exact ratio, and stays exact until
the moment it is converted to a float for output. That is a real design merit
and this paper says so.

**On injectivity: the module is wrong, or at least the label is.** Thirty named
tones take only twenty-three distinct values. Some collisions are defensible —
`electron` and `positron` are given the same tone, which is correct, since they
have the same mass. Others are not: `nu_mu` and `down` are both `3520/3` Hz, and
`nu_tau` and `charm` are both `1320` Hz. A listener cannot distinguish a muon
neutrino from a down quark, and neither can a decoder.

**On the durations: the label is false.** `tools.py` describes the rest
durations as *"rest durations in samples, exact integer arithmetic"* at
confidence `ESTABLISHED`. The arithmetic is integer, but it is not exact: the
durations use `//`, and four of six discard a fractional sample.

## Scope

**Claimed:** the exactness of the frequency ratios; the exact count of distinct
frequencies and the collision structure; the exact truncation error of each rest
duration; the ulp behaviour of the `ω = 2πf` round trip.

**Not claimed:** that any particular frequency assignment is *correct* in a
physical sense. Whether the Higgs "should" sound like A2 is not a question this
paper can settle and no code here tests it. This paper is about whether the
codec does what its own documentation says.
