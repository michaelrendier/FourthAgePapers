A twin pair (p, p+2) has one number between its members, the *centre* c = p+1. Two consecutive centres cut a *wheel* of circumference L = c₍ᵢ₊₁₎ − cᵢ, zero point at cᵢ; the end of one wheel is the beginning of the next.

## 2. One run, N = 10⁷

```python
r, c, L, ins = T.report(10**7)
for k, v in r.items(): print(f'{k}: {v}')
```

## 3. The wheel-size distribution against a random cutting (failed model kept)
If the 6-steps were cut at random with the same mean, wheel sizes m = L/6 would be geometric. They are not: m = 1 is rarer, and the singular series of the prime quadruplet (c−1, c+1, c+5, c+7) predicts by how much.

```python
m = L[1:] // 6; q = 1/m.mean(); obs = np.bincount(m)[1:]/len(m)
for k in range(1, 9): print(k, round(float(obs[k-1]),5), round(float(q*(1-q)**(k-1)),5), round(float(obs[k-1]/(q*(1-q)**(k-1))),3))
print('predicted ratio for m = 1:', r['predicted P(m=1)/geometric = S(0,2,6,8)/(6·S(0,2)²)'])
```

## 4. Brun's sum: how far the partial sum is from B₂
The tail should be 4C₂/ln N if the twin density is the Hardy–Littlewood one.

```python
print(r['Brun partial sum'], r['B₂ − partial sum'], r['4C₂/ln N'])
```
